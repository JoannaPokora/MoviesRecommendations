```python
import sys, os
project_root = r"C:/Users/Asia/Desktop/mocadr/MoviesRecommendations"
sys.path.insert(0, project_root)
import pandas as pd
from modules.utils import build_rating_matrix

Z, u, m = build_rating_matrix("C:/Users/Asia/Desktop/mocadr/MoviesRecommendations/data/ratings.csv")
```

```python
from sklearn.decomposition import TruncatedSVD
import numpy as np
from modules.predict import predict

svd = TruncatedSVD(n_components=min(Z.shape)-1, random_state=42)
svd.fit(Z)

cumvar = np.cumsum(svd.explained_variance_ratio_)
k = np.argmax(cumvar >= 0.90) + 1
print(k)

svd = TruncatedSVD(n_components=k, random_state=42)
svd.fit(Z)
Sigma2 = np.diag(svd.singular_values_)
VT = svd.components_
W = svd.transform(Z) / svd.singular_values_
H = np.dot(Sigma2, VT)
Z_approx = np.dot(W, H)
```

```python

predict([Z_approx, u, m], "C:/Users/Asia/Desktop/mocadr/MoviesRecommendations/data/sample_test.csv", "C:/Users/Asia/Desktop/mocadr/MoviesRecommendations/data/preds.csv")
```

```python
import sys, os
project_root = r"C:/Users/Asia/Desktop/mocadr/MoviesRecommendations"
sys.path.insert(0, project_root)

import pandas as pd
from sklearn.model_selection import KFold
from modules.predict import predict
from modules.utils import build_rating_matrix
from sklearn.decomposition import NMF
import numpy as np

df = pd.read_csv("C:/Users/Asia/Desktop/mocadr/MoviesRecommendations/data/ratings.csv")

folds = KFold(n_splits = 5, shuffle = True, random_state = 42)

n = 3
rss = {}
while n <= 13:
    print(n)
    rss_n = []
    for train_idx, test_idx in folds.split(df):
        train_df = df.iloc[train_idx].copy()
        test_df = df.iloc[test_idx].copy()

        Z_train, user_map_train, movie_map_train = build_rating_matrix(df = train_df)

        model = NMF(n_components = n, init = 'random', random_state = 0, max_iter = 1000)
        W = model.fit_transform(Z_train)
        H = model.components_
        Z_approx = np.dot(W, H)

        Z_pred = predict([Z_approx, user_map_train, movie_map_train], df = test_df.drop(columns=["rating"]))

        rss_n.append(np.sum((test_df["rating"] - Z_pred["rating"])**2))
    rss[n] = np.mean(rss_n)
    print(rss[n])
    n += 1
```
